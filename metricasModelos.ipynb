```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df_metricas = pd.read_excel("./bases/MetricasModelos6.xlsx", index_col=0)

modelos = ["Reg_Linear", "Rand_Forest", "Ex_Trees"]
nomes_modelos = ["Regressão Linear", "Random Forest", "Extra Trees"]

fig, axes = plt.subplots(1, 3, figsize=(16, 6))

metricas = {
    "MAE":  ("MAE_SLOG",  "MAE_CLOG"),
    "RMSE": ("RMSE_SLOG", "RMSE_CLOG"),
    "R²":   ("R2_SLOG",   "R2_CLOG")
}

x = np.arange(len(modelos))
largura = 0.35

for ax, (metrica, (sem_log, com_log)) in zip(axes, metricas.items()):
    valores_sem_log = df_metricas.loc[sem_log, modelos].values.astype(float)
    valores_com_log = df_metricas.loc[com_log, modelos].values.astype(float)

    # substitui valores absurdos por NaN para não distorcer o gráfico
    limite = 1e6
    valores_sem_log = np.where(valores_sem_log > limite, np.nan, valores_sem_log)
    valores_com_log = np.where(valores_com_log > limite, np.nan, valores_com_log)

    ax.bar(x - largura/2, valores_sem_log, width=largura, label="Sem Log", color="steelblue")
    ax.bar(x + largura/2, valores_com_log, width=largura, label="Com Log", color="darkorange")

    ax.set_xticks(x)
    ax.set_xticklabels(nomes_modelos, rotation=15, ha="right")
    ax.set_title(metrica)
    ax.set_ylabel(metrica)
    ax.legend()

plt.suptitle("Comparação de Desempenho dos Modelos", fontsize=14, fontweight="bold")
plt.tight_layout()
# plt.savefig("comparacao_modelos.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df_metricas = pd.read_excel("./bases/MetricasModelos6.xlsx", index_col=0)

modelos = ["Rand_Forest"]
nomes_modelos = ["Random Forest"]

fig, ax = plt.subplots(figsize=(3, 4))

# Extrair valores para R² Sem Log
r2_sem_log = df_metricas.loc["R2_SLOG", modelos].values.astype(float)

# Substituir valores absurdos por NaN
limite = 1e6
r2_sem_log = np.where(r2_sem_log > limite, np.nan, r2_sem_log)

x = np.arange(len(modelos))
largura = 0.4

# Plotar barra
barra = ax.bar(x, r2_sem_log, width=largura, color="steelblue")

# Adicionar valor da barra com 2 casas decimais
for b, valor in zip(barra, r2_sem_log):
    altura = b.get_height()
    ax.text(b.get_x() + b.get_width()/2., altura,
            f'{valor:.2f}',
            ha='center', va='bottom', fontsize=12, fontweight='bold')

ax.set_xticks(x)
ax.set_xticklabels(nomes_modelos)
ax.set_title("R² (Sem Log)", fontsize=12, fontweight="bold")
ax.set_ylabel("R²")
ax.set_ylim(0, max(r2_sem_log) * 1.15)

plt.suptitle("Desempenho - R²", fontsize=14, fontweight="bold")
plt.tight_layout()
# plt.savefig("comparacao_modelos.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
df_metricas = pd.read_excel("./bases/MetricasModelos6.xlsx", index_col=0)

modelos = ["Rand_Forest"]
nomes_modelos = ["Random Forest"]

fig, axes = plt.subplots(1, 3, figsize=(14, 5))

metricas = {
    "MAE":  "MAE_SLOG",
    "RMSE": "RMSE_SLOG",
    "R²":   "R2_SLOG"
}

x = np.arange(len(modelos))
largura = 0.5

for ax, (metrica, metrica_col) in zip(axes, metricas.items()):
    valores = df_metricas.loc[metrica_col, modelos].values.astype(float)
    
    # substitui valores absurdos por NaN para não distorcer o gráfico
    limite = 1e6
    valores = np.where(valores > limite, np.nan, valores)

    barras = ax.bar(x, valores, width=largura, color="steelblue")
    
    # Adicionar valores das barras com 2 casas decimais
    for barra, valor in zip(barras, valores):
        altura = barra.get_height()
        ax.text(barra.get_x() + barra.get_width()/2., altura,
                f'{valor:.2f}',
                ha='center', va='bottom', fontsize=10, fontweight='bold')

    ax.set_xticks(x)
    ax.set_xticklabels(nomes_modelos)
    ax.set_title(metrica, fontsize=12, fontweight="bold")
    ax.set_ylabel(metrica)

plt.suptitle("Comparação de Desempenho - Random Forest (Sem Log)", fontsize=14, fontweight="bold")
plt.tight_layout()
# plt.savefig("comparacao_modelos.png", dpi=300, bbox_inches="tight")
plt.show()
```
